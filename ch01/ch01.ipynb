#### EXAMPLE 1.4 An order-of-magnitude estimate
page 38

**EVALUATE** Clearly your novel needs rewriting. Maybe your hero could
be satisfied with 1 million dollars’ worth of gold? We have seen that the
mass of gold in this case is about 25 kilograms, an amount which your
hero should be able to carry and which would easily fit in a briefcase.
If you want a more spectacular amount, try the calculation again with
a suitcase full of five-carat (1-gram) diamonds, each worth $500,000.
Would this work?

```python
# 1. Define custom currency unit
USD = u.def_unit("USD")

# 2. Define given parameters
price_per_diamond = 500_000 * USD
mass_per_diamond = 1.0 * u.g  # 5 carats = 1 gram
total_value = 1e9 * USD
density_diamond = 3.51 * u.g / (u.cm**3)  # Standard density of diamond

# 3. Calculate Mass
# Number of diamonds = total_value / price_per_diamond
num_diamonds = total_value / price_per_diamond
total_mass = (num_diamonds * mass_per_diamond).to(u.kg)

# 4. Calculate Volume
total_volume = (total_mass / density_diamond).to(u.l)

# Output results
print(f"Total Mass: {total_mass}")
print(f"Total Volume: {total_volume:.2f}")
```

### Scalar Product (Dot Product)

We denote the **scalar product** of two vectors $\vec{A}$ and $\vec{B}$ by $\vec{A} \cdot \vec{B}$.

Because of this notation, the scalar product is also called the **dot product**. Although $\vec{A}$ and $\vec{B}$ are vectors, the quantity $\vec{A} \cdot \vec{B}$ is a scalar.


<img src="..//pics/Fig_1_27.png" width=300> 

Find the scalar product $\vec{A} \cdot \vec{B}$ of the two vectors in Fig. 1.27. The magnitudes
of the vectors are A = 4.00 and B = 5.00.

```python
A_theta = vp.radians(53)
B_theta = vp.radians(130)

# Using vector components
A = vp.vec(4.00 * vp.cos(A_theta), 4.00 * vp.sin(A_theta), 0)
B = vp.vec(5.00 * vp.cos(B_theta), 5.00 * vp.sin(B_theta), 0)

AB_dot = vp.dot(A, B)
print(f"A B dot using components: {AB_dot:.2f}")

# Using dot equation
phi = B_theta - A_theta
AB_dot = 4 * 5 * vp.cos(phi)
print(f"A B dot using equation: {AB_dot:.2f}")
```

## EXAMPLE 1.10 Finding an angle with the scalar product

<img src="..//pics/Fig_1_28.png" width=400> 

Find the angle between the vectors
$$\vec{A} = (2.00 \hat{i} + 3.00 \hat{j} + 1.00 \hat{k}$$

$$\vec{B} = (-4.00 \hat{i} + 2.00 \hat{j} - 1.00 \hat{k}$$


```python
# calculate the angle phi between two vectors A y B
A = vp.vec(2, 3, 1)
B = vp.vec(-4, 2, -1)

# getting angle phi by equation
# dot_AB = mag(A) * mag(B) * cos(phi) ==> phi = acos(dot_AB/(mag(A) * mag(B)))
dot_AB = vp.dot(A, B)

phi_rad = vp.acos(dot_AB / (A.mag * B.mag))
phi_deg = (phi_rad * u.rad).to(u.deg)
phi_deg
print(f"Angle phi = {phi_deg:.0f}")

# using the built-in angle function in vPython
phi_rad_vPython = vp.diff_angle(A, B) * u.rad
phi_deg_vPython = phi_rad_vPython.to(u.deg)

print(f"Angle phi (vPython) = {phi_deg_vPython:.0f}")
```

### Vector Product

<img src="..//pics/Fig_1_29.png" width=300> 

We denote the **vector product** of two vectors $\vec{A}$ and $\vec{B}$, also called the **cross product**, by  $\vec{A} \times \vec{B}$. As the name suggests, the vector product is itself a vector. We’ll use this product in Chapter 10 to describe torque and angular momentum; in Chapters 27 and 28 we’ll use it to describe magnetic fields and forces.

To define the vector product  $\vec{A} \times \vec{B}$, we again draw the two vectors $\vec{A}$ and $\vec{B}$ with their tails at the same point (Fig. 1.29a). The two vectors then lie in a plane. We define the vector product to be a vector quantity with a direction perpendicular to this plane
(that is, perpendicular to both $\vec{A}$ and $\vec{B}$) and a magnitude equal to $AB \sin \phi$

The result is a vector that is opposite to the vector $\vec{A} \times \vec{B}$. The vector product is not commutative but instead is anticommutative: For any two vectors $\vec{A}$ and $\vec{B}$,

$$\vec{A} \times \vec{B} = - \vec{A} \times \vec{B}$$

## EXAMPLE 1.11 Calculating a vector product
page 52

<img src="..//pics/Fig_1_32.png" width=400> 

Vector  $\vec{A}$ has magnitude 6 units and is in the direction of the +x-axis.
Vector  $\vec{B}$ has magnitude 4 units and lies in the xy-plane, making an
angle of 30° with the +x-axis (Fig. 1.32). Find the vector product
 $\vec{C} = \vec{A} \times \vec{B}$

```python
phi = vp.radians(30)
A = vp.vec(6, 0, 0)
B = vp.vec(4 * vp.cos(phi), 4 * vp.sin(phi), 0)

# C_mag = mag(A) * mag(B) * sin(phi)
C_mag = A.mag * B.mag * vp.sin(phi)

# By the right-hand rule, vector C points to the z-axis
C = vp.vec(0, 0, C_mag)
print(f"cross product C = {fmt_vec(C)}")

# using the built-in cross product function in vPython

C_vpython = vp.cross(A, B)
print(f"(Built in cross), C = {fmt_vec(C_vpython)}")
```

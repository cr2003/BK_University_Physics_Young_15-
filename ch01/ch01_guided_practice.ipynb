#### VP1.7.1

Consider the three vectors $\vec{A}$, $\vec{B}$ and $\vec{C}$ in Example 1.7.
If a fourth vector $\vec{D}$ is added  $\vec{A} + \vec{B} + \vec{C}$, the result is zero:
 $\vec{A} + \vec{B} + \vec{C} + \vec{D} = 0$.
 
Find the magnitude and direction of $\vec{D}$ . State
the direction of $\vec{D}$ in terms of an angle measured counterclockwise from
the positive x-axis, and state in which quadrant this angle lies.

```python
### Example 1.7# --- 1. Define Vectors ---

# Vector A: 72.4 m at 32.0 deg from +y axis (58.0 deg from +x axis)
angle_A = vp.radians(90 - 32.0)
A = vp.vec(72.4 * vp.cos(angle_A), 72.4 * vp.sin(angle_A), 0)

# Vector B: 57.3 m pointing 36.0 deg below the negative x-axis (180 + 36 deg)
angle_B = vp.radians(180 + 36.0)
B = vp.vec(57.3 * vp.cos(angle_B), 57.3 * vp.sin(angle_B), 0)

# Vector C: 17.8 m pointing straight down (-y direction)
C = vp.vec(0, -17.8, 0)

# --- 2. Calculate Resultant Vector R --
R = A + B + C

# --- 3. Compute Magnitude and Angle ---
R_mag = vp.mag(R)

# Angle theta measured counterclockwise from the +x axis
theta_deg = vp.degrees(vp.atan2(R.y, R.x))
if theta_deg < 0:
    theta_deg += 360

# --- 4. Output Results ---
print(f"Vector R: {fmt_vec(R)} m")
print(f"Magnitude |R|: {R_mag:.1f} m")
print(f"Angle θ: {theta_deg:.0f}°")
```

```python
### According to the geometry, vector D = -R

D = -R
print(f"Check that A + B + C + D = {A + B + C + D}")

# Compute Magnitude and Angle ---
D_mag = D.mag

# Angle theta measured counterclockwise from the +x axis
theta_deg = vp.degrees(vp.atan2(D.y, D.x))
if theta_deg < 0:
    theta_deg += 360

# --- Output Results ---
print(f"Vector D: {fmt_vec(D)} m")
print(f"Magnitude |D|: {D_mag:.1f} m")
print(f"Angle θ: {theta_deg:.0f}°")
```

#### VP1.10.2 
Vector $\vec{C}$ has magnitude 6.50 and is at an angle of 55.0° measured
counterclockwise from the +x-axis toward the +y-axis. Vector $\vec{D}$
has components Dx = +4.80 and Dy = -8.40. 

**(a)** Calculate the scalar product $\vec{C} \cdot \vec{D}$ . 

**(b)** Find the angle $\phi$ between the vectors $\vec{C}$ and $\vec{D}$ .

```python
print("(a)")
# Vector C: 6.50 at 55.0° standard
angle_C = vp.radians(55.0)
C = vp.vec(6.50 * vp.cos(angle_C), 6.50 * vp.sin(angle_C), 0)
print(f"vector C: {fmt_vec(C)}")

D = vp.vec(4.80, -8.40, 0)
print(f"vector D: {fmt_vec(D)}")


CD_dot = vp.dot(C, D)
CD_dot
print(f"Scalar product CD_dot = {CD_dot:.1f}°")
```

```python
print("(b)")

print("solve by equation")
# CD_dot = C.mag * D.mag * cos(phi) ==> phi = (acos((CD_dot)/((C.mag * D.mag))
phi = vp.degrees((vp.acos((CD_dot) / (C.mag * D.mag))))
print(phi)

print("solve by vPython built-in function")
phi = vp.degrees(vp.diff_angle(C, D))
print(phi)

print(f"Angle phi = {phi:.0f}°")
```

#### VP1.10.4 

If a force $\vec{F}$ acts on an object as that object moves through
a displacement $\vec{s}$, the work done by that force equals the scalar product
of $\vec{F}$ and $\vec{s}$: $\vec{W} = \vec{F} \cdot \vec{s}$. 

A certain object moves through displacement
$\vec{s} = (4.00 m)\hat{i} + (5.00 m)\hat{j}$. As it moves it is acted on by force
$\vec{F}$ which has x-component Fx = -12.0 N (1 N = 1 newton is the
SI unit of force). The work done by this force is $26.0 N \cdot m = 26.0 J$
(1 J = 1 joule = 1 newton-meter is the SI unit of work). 

**(a)** Find the y-component of $\vec{F}$. 

**(b)** Find the angle between $\vec{F}$ and $\vec{s}$.



```python
print("(a)")
s_x = 4.00 * u.m
s_y = 5.00 * u.m
F_x = -12.0 * u.N
W = 26.0 * u.J

# W = F . S
# W = (F_x * s_x) + (F_y * s_y) ==> F_y = (W - (F_x * s_x)) / s_y
F_y = (W - (F_x * s_x)) / s_y
F_y.to(u.N)
```

```python
print("(b)")
# Calculate phi by equation and components

# 1. calculate the magnitude of s
s_mag = vp.sqrt(s_x.value**2 + s_y.value**2)

# 2. calculate the magnitude of F
F_mag = vp.sqrt(F_x.value**2 + F_y.value**2)

# 3. calculate cos(phi) and phi
# W = F_mag * s_mag * cos(phi)
cos_phi = W.value / (F_mag * s_mag)
phi = vp.degrees(vp.acos(cos_phi))
print(f"phi by equation: {phi}")


# Calculate phi by vector functions
F = vp.vec(-12.0, 14.8, 0)
s = vp.vec(4.00, 5.00, 0)
phi = vp.degrees(vp.diff_angle(F, s))
print(f"phi by vector functions: {phi}")
print(f"Rounded by 3 significant figures phi = {phi:.1f}°")
```

An air-conditioning unit is fastened to a roof that slopes at an angle of
35° above the horizontal (Fig. 1.33). Its weight is a force  $\vec{F}$ on the air
conditioner that is directed vertically downward. In order that the unit
not crush the roof tiles, the component of the unit’s weight perpendicular
to the roof cannot exceed 425 N. (One newton, or 1 N, is the SI unit
of force.) 

**(a)** What is the maximum allowed weight of the unit? 

**(b)** If
the fasteners fail, the unit slides 1.50 m along the roof before it comes
to a halt against a ledge. How much work does the weight force do on
the unit during its slide if the unit has the weight calculated in part (a)?
The work done by a force  $\vec{F}$ on an object that undergoes a displacement
 $\vec{F}$ is $W =  \vec{F} \cdot  \vec{s}$.

```python
# print("(a)")

theta = vp.radians(35)
F_y = 425 * u.N
# F_y = F cos(theta) ==> F = F_y / cos(theta)
F = F_y / vp.cos(theta)
print(f"Max. allowed weight F = {F:.0f}")
```

```python
print("(b)")

delta_x = 1.50 * u.m
F_x = F * vp.sin(theta)
W = F_x * delta_x
W.to(u.J)
print(f"W = {W.to(u.J):.0f}")
```

#### 1.40

•• **(a)** Find the scalar product of the vectors $\vec{A}$ and $\vec{B}$ given in
Exercise 1.36. 

NB: From exercise 1.36 vectors are:

$\vec{A} = 4.00\hat{i} + 7.00\hat{j}$

$\vec{B} = 5.00\hat{i} - 2.00\hat{j}$


```python
A = vp.vec(4, 7, 0)
B = vp.vec(5, -2, 0)
dot_A_B = vp.dot(A, B)
dot_A_B
```

**(b)** Find the angle between these two vectors.

```python
phi = vp.degrees(vp.diff_angle(A, B))
print(f"Angle phi = {phi:.1f}°")
```

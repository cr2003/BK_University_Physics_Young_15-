#### Exercise 1.1

How many years older will you be 1.00 gigasecond from now?
(Assume a 365-day year.)

```python
print("1. Using AstroPy units")
delta_time = 1 * u.gigasecond
print(f"time_years: {delta_time.to(u.yr):.1f}")

# Manual operation
print("\n2. Using factors")

# delta_time_s = 1Gs = 10^9 s
delta_time_s = 1e9

print(f"delta_time_s: {delta_time_s:,.0f}")
# years = (delta_time_s s) * (1 min/60 s) * (1 hr/60 min) * (1 day/24 hr) * (1 yr/365 day)
years = (delta_time_s) * (1 / 60) * (1 / 60) * (1 / 24) * (1 / 365)
print(f"time_years: {years:,.1f}")
```

#### Exercise 1.2
You read in a brochure that Kensington Gardens, one of London’s
eight Royal Parks, covers an area of 265 acres. An acre is a British unit
widely used in land measurement in the UK (1 acre = 43,560 ft2 and
1 ft = 0.3048 m, see Appendix C). What is the area of Kensington Gardens
in square meters? in square kilometers? in hectares (a hectare is a unit of
area equal to 10,000 m2 used in land measurement)?

```python
area_acres = 265 * u.imperial.ac
print(f"area acres: {area_acres:.0f}")

area_meters = area_acres.to(u.m**2)
print(f"area meters: {area_meters:,.0f}")

area_km = area_acres.to(u.km**2)
print(f"area kilometers: {area_km:,.2f}")

area_hectares = area_acres.to(u.ha)
print(f"area hectares: {area_hectares:,.0f}")
```

#### 1.6 
• The speed limits in Ireland were changed from imperial (miles
per hour, mi/h) to metric (kilometers per hour, km/h) on 20 January
2005. Use 1 mi = 5280 ft and 1 ft = 30.48 cm to: 

**(a)** convert the former
30 mi/h speed limit for build-up areas to units of km/h and compare
this result with the new 50 km/h speed limit; 

**(b)** convert the new
120 km/h speed limit for motorways to units of mi/h and compare this
result with the old 70 mi/h speed limit.

```python
print("(a)")
limit_bup_mih = 30 * (u.imperial.mi / u.hr)
new_limit_kmh = 50 * (u.km / u.hr)

translate_kmh = limit_bup_mih.to((u.km / u.hr))
delta_limit_kmh = new_limit_kmh - translate_kmh

print(f"limit in mi/hr = {limit_bup_mih}")
print(f"direct translate limit in km/hr = {translate_kmh:.1f}")
print(f"new limit {new_limit_kmh}")
print(f"increase in limit: {delta_limit_kmh:.1f}")
```

```python
print("(b)")
limit_mtw_mih = 70 * (u.imperial.mi / u.hr)
new_limit_kmh = 120 * (u.km / u.hr)

translate_kmh = limit_mtw_mih.to((u.km / u.hr))
delta_limit_kmh = new_limit_kmh - translate_kmh

print(f"limit in mi/hr = {limit_mtw_mih}")
print(f"direct translate limit in km/hr = {translate_kmh:.1f}")
print(f"new limit {new_limit_kmh}")
print(f"increase in limit: {delta_limit_kmh:.1f}")
```

#### 1.7 
• Fuel consumption of a car is the amount of fuel used per unit
distance. It is customary to give fuel consumption in liters per 100 km
(L/100 km). For example, the 2019 Austin Mini Cooper has a fuel consumption
of 5.5 L/100 km, that is, it uses 5.5 liters of petrol per every
100 km driven. 

**(a)** If this car’s petrol tank holds 40 L, how many tanks
of petrol will you use to drive 1200 km? 

**(b)** You read in an English car
magazine that the classic 1964 Austin Mini Cooper has an average estimated
“mileage of 30.5 miles per gallon” (where 1 mile = 1.609 km is
a British unit of distance and 1 gallon (UK) = 4.546 L is a British unit
of capacity, see Appendix C). Find the fuel consumption of this car and
compare it with the 2019 model.


```python
# --- Setup Custom UK Gallon ---
# Define UK gallon as explicitly stated in the problem (1 UK gal = 4.546 L)
uk_gal = u.def_unit("uk_gal", 4.546 * u.L)

# --- Given Data ---
consumption_2019 = 5.5 * (u.L / (100 * u.km))
tank_capacity = 40 * u.L
distance_a = 1200 * u.km
mileage_1964 = 30.5 * (u.imperial.mi / uk_gal)

print("--- PART A ---")
# 1. Calculate total liters needed for 1200 km
total_fuel_needed = distance_a * consumption_2019
print(f"Total fuel needed: {total_fuel_needed:.1f}")

# 2. Calculate the number of tanks
tanks_needed = total_fuel_needed / tank_capacity
print(f"Number of tanks required: {tanks_needed.value:.2f} tanks")


print("\n--- PART B ---")
# 1. Invert the mileage to get gallons per mile
gal_per_mile = 1 / mileage_1964

# 2. Convert to liters per kilometer using astropy's equivalencies/conversion
# Since 1 mi = 1.609 km is standard in astropy, we can convert directly
L_per_km_1964 = gal_per_mile.to(u.L / u.km)

# 3. Scale to L / 100 km
consumption_1964 = L_per_km_1964 * 100
print(f"1964 model fuel consumption: {consumption_1964.value:.2f} L/100 km")
print(f"2019 model fuel consumption: {consumption_2019.to(u.L / (100 * u.km)):.2f}")


# 4. Compare
difference = consumption_1964.value - 5.5
print(
    f"The 1964 classic uses {difference:.2f} more Liters per 100 km than the 2019 model."
)
```

**(a)** The recommended daily allowance (RDA) of the trace
metal magnesium is $ 410 mg/day $ for males. Express this quantity in
$\mu g/day$. 

```python
RDA_mg = 410 * (u.mg / u.day)

RDA_ug = RDA_mg.to(u.ug / u.day)

print(f"RDA magnesium: {RDA_mg:.0f} = {RDA_ug:,.0f}")
```

**(b)** For adults, the RDA of the amino acid lysine is 12 mg per
kg of body weight. How many grams per day should a 77-kg adult receive?

```python
RDA_mg = 12 * (u.mg / u.kg)
adult_77 = 77 * u.kg

dose = RDA_mg * adult_77

print(f"RDA amino acid lysine per day: {dose.to(u.g):.2f}")
```

**(c)** A typical multivitamin tablet can contain 2.0 mg of vitamin
B2 (riboflavin), and the RDA is 0.0030 g/day. How many such tablets
should a person take each day to get the proper amount of this vitamin,
if he gets none from other sources? 

```python
# --- Setup Custom Unit ---
tablet = u.def_unit("tablet")

# --- Given Data ---
b2_per_tablet = 2.0 * (u.mg / tablet)
rda = 0.0030 * (u.g / u.day)

# --- Calculation ---
# Divide RDA by the amount per tablet to get tablets per day
tablets_per_day = rda / b2_per_tablet

# Convert grams to milligrams to get a dimensionless count of tablets per day
result = tablets_per_day.to(tablet / u.day)

print(f"Required intake: {result:.1f}")
```

**(d)** The RDA for the trace element
selenium is 0.000070 g/day. Express this dose in mg/day.

```python
RDA_gd = 0.000070 * (u.g / u.d)

RDA_mgd = RDA_gd.to(u.mg / u.d)
print(f"Required intake: {RDA_mgd:.2f}")
```

#### 1.9 
•• Neptunium. In the fall of 2002, scientists at Los Alamos
National Laboratory determined that the critical mass of neptunium-237
is about 60 kg. The critical mass of a fissionable material is the minimum
amount that must be brought together to start a nuclear chain reaction.
Neptunium-237 has a density of 19.5 g/cm3. What would be the
radius of a sphere of this material that has a critical mass?

```python
mass = (60 * u.kg).to(u.g)

rho = 19.5 * (u.g / u.cm**3)

# rho = m / V ==> V = m / rho
V = mass / rho

# volume of sphere: 4/3 pi r^3 ==> r = cbrt(3 * V / 4 * pi)
r = np.cbrt((3 * V) / (4 * np.pi))
print(f"the radius = {r:.1f}")
```

Bacteria vary in size, but a diameter of $2.0 \mu m$
is not unusual. What are the volume (in cubic centimeters) and surface
area (in square millimeters) of a spherical bacterium

```python
r = (2.0 / 2) * u.um

V = (4 / 3) * np.pi * r**3
A = 4 * np.pi * r**2
print(f"Bacterium volume={V.to(u.cm**3):.1e}")
print(f"Bacterium area={A.to(u.mm**2):.1e}")
```

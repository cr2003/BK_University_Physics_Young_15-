#### Exercise 1.11

• With a wooden ruler, you measure the length of a rectangular
piece of sheet metal to be 12 mm. With micrometer calipers, you measure
the width of the rectangle to be 5.98 mm. Use the correct number
of significant figures: What are 

**(a)** the area of the rectangle; 

```python
print("(a)")
l = 12 * u.mm
w = 5.98 * u.mm

A = l * w
print(f"Area = {A:.0f}")
```

**(b)** the ratio
of the rectangle’s width to its length; 

```python
print("(b)")

ratio = w / l
print(f"ratio = {ratio:.2f}")
```

**(c)** the perimeter of the rectangle;

```python
perimeter = (2 * l) + (2 * w)
print(f"perimeter = {perimeter:.0f}")
```

**(d)** the difference between the length and the width; and 

```python
d = l - w
print(f"difference = {d:.0f}")
```

(e) the ratio of the length to the width?

```python
ratio_lw = l / w
print(f"ratio_lw = {ratio_lw:.1f}")
```

#### 1.12
• The volume of a solid cylinder is given by $V = \pi r^2 h$, where r is
the radius and h is the height. You measure the radius and height of a thin
cylindrical wire and obtain the results r = 0.036 cm and h = 12.1 cm.
What do your measurements give for the volume of the wire in $mm^3$?
Use the correct number of significant figures in your answer.

```python
r = 0.036 * u.cm  # 2 significant figures
h = 12.1 * u.cm  # 3 significant figures
V = np.pi * r**2 * h
V = V.to(u.mm**3)
print(f"Volume = {V:.0f}")  # limit to 2 significant figures
```

#### 1.13 
•• A useful and easy-to-remember approximate value for the
number of seconds in a year is $\pi \times 10^7$. Determine the percent error in
this approximate value. (There are 365.24 days in one year.)

```python
# --- Given Data ---
# 365.24 days has 5 significant figures
actual_year = 365.24 * u.day
approx_seconds = np.pi * 1e7 * u.s

# --- Conversion ---
# Convert the actual year duration directly into seconds
actual_seconds = actual_year.to(u.s)

# --- Percent Error Calculation ---
# Percent Error = (|Approximate - Actual| / Actual) * 100
error_fraction = np.abs(approx_seconds - actual_seconds) / actual_seconds
percent_error = error_fraction * 100

# --- Print Results ---
# The value 365.24 has 5 sig figs, but pi is an approximation here.
# Keeping it to 2 decimal places matches common textbook answers.
print(f"Actual seconds in a year:      {actual_seconds.value:.0f} s")
print(f"Approximate seconds (\\pi*10^7): {approx_seconds.value:.0f} s")
print(f"Percent error:                 {percent_error.value:.2f}%")
```

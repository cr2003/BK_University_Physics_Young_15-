#### 1.16 
• How many liters of petrol are used in Italy in one day? Assume
that there are two cars for every three people, that each car is driven an
average of 10,000 kilometers per year, and that the petrol consumption
of an average car is 8 liters per 100 kilometers. Assume that the population
of Italy is approximately 60 million.

```python
num_cars = 60e6 * (2 / 3)
car_driven_yr = 10_000 * (u.km / u.yr)
car_driven_day = car_driven_yr.to(u.km / u.d)
km_day = car_driven_day * num_cars
consumption = (8 * u.l) / (100 * u.km)
liters = km_day * consumption
print(f"Italy car petrol: {liters:,.0f}")
```

```python
# --- Setup Custom Counting Units ---
car = u.def_unit("car")
person = u.def_unit("person")

# --- Given Data ---
population = 60_000_000 * person
car_ratio = (2 / 3) * (car / person)
distance_per_year = 10_000 * (u.km / (car * u.yr))
consumption = 8 * (u.L / (100 * u.km))

# --- Calculations ---
# 1. Total number of cars in Italy
total_cars = population * car_ratio

# 2. Total distance driven by all cars combined per year
total_distance_per_year = total_cars * distance_per_year

# 3. Total liters of petrol consumed per year
total_fuel_per_year = total_distance_per_year * consumption

# 4. Convert total fuel consumption per year to consumption per day
# AstroPy handles the year-to-day time translation automatically
total_fuel_per_day = total_fuel_per_year.to(u.L / u.day)

# --- Print Results ---
print(f"Total cars in Italy: {total_cars:,.0f}")
print(f"Total petrol used per day: {total_fuel_per_day:,.0f}")


# Since this is a Fermi estimation problem, it is best expressed in millions of liters:
liters_in_millions = total_fuel_per_day.to(u.ML / u.d)

print(f"Estimated daily consumption in millions: ~{liters_in_millions:.0f} ")
```

#### 1.17 
• In Wagner’s opera Das Rheingold, the goddess Freia is ransomed
for a pile of gold just tall enough and wide enough to hide her
from sight. Estimate the monetary value of this pile. The density of gold
is 19.3 g/cm3, and take its value to be about $40 per gram.


```python
import astropy.units as u

# --- Setup Custom Monetary Unit ---
# AstroPy doesn't have a currency namespace, so we treat USD as a dimensionless custom unit.
usd = u.def_unit("USD")

# --- 1. Define Estimation Dimensions ---
height = 1.7 * u.m
width = 0.5 * u.m
depth = 0.5 * u.m

# --- 2. Given Problem Constants ---
density_gold = 19.3 * (u.g / (u.cm**3))
value_per_gram = 40.0 * (usd / u.g)

# --- 3. Calculations ---
# Volume of the rectangular box
volume = height * width * depth

# Convert volume to cm^3 so it cancels perfectly with the density unit
volume_cm3 = volume.to(u.cm**3)

# Mass = Volume * Density
mass_gold = volume_cm3 * density_gold

# Total monetary value = Mass * Price per gram
total_value = mass_gold * value_per_gram

# --- 4. Print Results ---
print(f"Estimated Volume: {volume_cm3:,.0f}")
print(f"Estimated Mass:   {mass_gold.to(u.kg):,.0f} ")
print(f"Estimated Value:  ${total_value:,.0f} ")

# Round to the nearest hundred million ($100,000,000)
rounded_value = round(total_value / 1e8) * 1e8

print(f"Estimated Rounded Value:  ${rounded_value:,.0f}")
```

#### 1.18 
• BIO Four astronauts are in a spherical space station. 

**(a)** If, as is typical, each of them breathes about 500 cm3 of air with each breath,
approximately what volume of air (in cubic meters) do these astronauts
breathe in a year? 

**(b)** What would the diameter (in meters) of the space
station have to be to contain all this air?

```python
# --- Setup Custom Counting Units ---
breath = u.def_unit("breath")
astronaut = u.def_unit("astronaut")

# --- Given & Estimated Data ---
num_astronauts = 4 * astronaut
volume_per_breath = 500 * ((u.cm) ** 3 / breath)
breaths_per_minute = 15 * (breath / u.min)

# --- Part (a): Total Volume Breathed in a Year ---
# Calculate total volume per year (AstroPy handles minute-to-year time translation)
total_volume_year = (
    num_astronauts.value * volume_per_breath * breaths_per_minute * (1 * u.yr)
)

# Convert to cubic meters (m^3)
volume_in_m3 = total_volume_year.to(u.m**3)

print("--- PART (a) ---")
print(f"Total volume of air breathed in a year: {volume_in_m3.value:,.0f} m^3")


# --- Part (b): Space Station Diameter ---
# Volume of a sphere: V = (4/3) * pi * r^3
# Radius: r = ((3 * V) / (4 * pi))^(1/3)
radius = ((3.0 * volume_in_m3) / (4.0 * np.pi)) ** (1.0 / 3.0)
diameter = 2.0 * radius

print("\n--- PART (b) ---")
print(f"Required diameter of the space station: {diameter.value:.1f} meters")
```

#### 1.19 
• You are using water to dilute small amounts of chemicals in the
laboratory, drop by drop. How many drops of water are in a 1.0 L bottle?
(Hint: Start by estimating the diameter of a drop of water.)

```python
diameter = 2 * u.mm
radius = diameter / 2

# Calculate the volume of a drop
# Volume of a sphere: V = (4/3) * pi * r^3
V = (4 / 3) * np.pi * radius**3

drops = 1 * u.l / V.to(u.l)
print(f"Number of drops: {drops:,.0f}")
# Round to the nearest hundred thousand (100,000)
rounded_value = round(drops / 1e5) * 1e5
print(f"Rounded number of drops: {rounded_value:,.0f}")
```

#### 1.20 
• BIO How many times does a human heart beat during a person’s
lifetime? How many liters of blood does it pump? (Estimate that the heart
pumps 50 cm3 of blood with each beat and assume a 365-day year.)

```python
lifetime = (75 * u.yr).to(u.min)
hearbeat_min = 70 * (1 / u.min)
pump = 50 * (u.cm**3)

# calculate the beats
total_beats = lifetime * hearbeat_min
# Round to the nearest 10^9
rounded_total_beats = round(total_beats / 1e9) * 1e9
print(f"Rounded total beats: {rounded_total_beats:.0e}")


# calculate the liter pumps
total_liters = (total_beats * pump).to(u.l)

# Round to the nearest 10^8
rounded_total_liters = round(total_liters / 1e8) * 1e8
print(f"Rounded total liters: {rounded_total_liters:.0e}")
```

```python
beat = u.def_unit("beat")
lifespan = (75 * u.yr).to(u.min)
hear_rate = 70 * (beat / u.min)
volume_per_beat = 50 * (u.cm**3 / beat)

# --- Part 1: Total Lifetime Heartbeats ---
# calculate the beats
total_beats = lifetime * hearbeat_min
lifetime_beats = lifespan * hear_rate

print("--- PART 1: LIFETIME BEATS ---")
print(f"Total heartbeats over a lifespan: {lifetime_beats}")
# Round to the nearest 10^9
rounded_total_beats = round(lifetime_beats / 1e9) * 1e9
print(f"Rounded total beats: {rounded_total_beats:.0e}")

# --- Part 2: Total Volume Pumped ---
# calculate the liter pumps
total_volume_cm3 = lifetime_beats * volume_per_beat

# Convert from cubic centimeters (cm^3) to Liters (L)
total_volume_liters = total_volume_cm3.to(u.L)

print("\n--- PART 2: BLOOD VOLUME PUMPED ---")
print(f"Total volume pumped: {total_volume_liters:,.0f}")

# Round to the nearest 10^8
rounded_total_liters = round(total_volume_liters / 1e8) * 1e8
print(f"Rounded total liters: {rounded_total_liters:.0e}")
```

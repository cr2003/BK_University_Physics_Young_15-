#### 1.21 

<img src="..//pics/Fig_E1_21.png" width=400> 

•• A postal employee drives a delivery truck along the route
shown in Fig. E1.21. Determine the magnitude and direction of the
resultant displacement by drawing a scale diagram. (See also Exercise
1.28 for a different approach.)



```python
A = vp.vec(0, 2.6, 0)  # km
B = vp.vec(4.0, 0, 0)  # km
theta = vp.radians(45)
C = vp.vec(3.1 * vp.cos(theta), 3.1 * vp.sin(theta), 0)  # km
displacement = A + B + C

displacement_mag = displacement.mag
phi_rad = np.atan2(displacement.y, displacement.x)
phi_deg = vp.degrees(phi_rad)
print(f"vector displacement: {fmt_vec(displacement, 1)}")
print(f"displacement <{displacement_mag:.1f} km, {phi_deg:.1f} deg>")
```

#### Ploting result

```python
# --- 1. Compute Data Vector Components Analytically ---
# Vector A: 2.6 km North
Ax, Ay = 0.0, 2.6
# Vector B: 4.0 km East
Bx, By = 4.0, 0.0
# Vector C: 3.1 km at 45 degrees North of East
angle = np.radians(45)
Cx, Cy = 3.1 * np.cos(angle), 3.1 * np.sin(angle)

# Calculate correct starting origins for tip-to-tail chaining
x_starts = [0, Ax, Ax + Bx]
y_starts = [0, Ay, Ay + By]

# End points
Rx = Ax + Bx + Cx
Ry = Ay + By + Cy

# --- 2. Matplotlib Setup & Sketching ---
plt.figure(figsize=(8, 6), dpi=100)
plt.grid(True, linestyle="--", alpha=0.6)

# Plot path vectors using correct starting origins
plt.quiver(
    x_starts,
    y_starts,
    [Ax, Bx, Cx],
    [Ay, By, Cy],
    angles="xy",
    scale_units="xy",
    scale=1,
    color=["#1f77b4", "#ff7f0e", "#2ca02c"],
    label="Route Path",
)

# FIX: Added raw string r'...' to prevent the \m mathbf syntax warning
plt.quiver(
    0,
    0,
    Rx,
    Ry,
    angles="xy",
    scale_units="xy",
    scale=1,
    color="purple",
    linestyle="--",
    alpha=0.8,
    label=r"Resultant ($\mathbf{R}$)",
)

# Add markers for start, turn joints, and end points
plt.scatter([0, Ax, Ax + Bx, Rx], [0, Ay, Ay + By, Ry], color="black", zorder=5)
plt.text(0, -0.3, "START", color="red", fontweight="bold", ha="center")
plt.text(
    Rx,
    Ry + 0.2,
    f"STOP\n({Rx:.1f}k, {Ry:.1f}j)",
    color="magenta",
    fontweight="bold",
    ha="center",
)

# Annotate individual vector labels near their paths
plt.text(Ax - 0.4, Ay / 2, "2.6 km\nNorth", color="#1f77b4", va="center")
plt.text(Ax + Bx / 2, Ay - 0.4, "4.0 km East", color="#ff7f0e", ha="center")
plt.text(
    Ax + Bx + Cx / 2 - 0.3,
    Ay + Cy / 2 + 0.3,
    "3.1 km\n@ 45°",
    color="#2ca02c",
    ha="center",
)

# Graph limits and adjustments
plt.xlim(-1, 8)
plt.ylim(-1, 6)
plt.axhline(0, color="black", linewidth=0.8, alpha=0.5)
plt.axvline(0, color="black", linewidth=0.8, alpha=0.5)

# Axis titles and labels
plt.title("Vector Displacement Scale Diagram (Problem 1.21)", fontsize=12, pad=10)
plt.xlabel("East / West (km)", fontsize=10)
plt.ylabel("North / South (km)", fontsize=10)
plt.legend(loc="upper left")
plt.gca().set_aspect("equal", adjustable="box")

plt.show()
```

#### 1.23 
•• A spelunker is surveying
a cave. She follows a passage 180
m straight west, then 210 m in a direction
45° east of south, and then
280 m at 30° east of north. After a
fourth displacement, she finds herself
back where she started. Use a scale drawing to determine the magnitude
and direction of the fourth displacement. (See also Problem 1.57
for a different approach.)

```python
# --- 1. Define Vectors ---

# Vector A: 180 m at 180 deg from +y axis (straight west)
A = vp.vec(-180, 0, 0)

# Vector B: 210 m pointing 45 deg east of south (270+45)
angle_B = vp.radians(270 + 45)
B = vp.vec(210 * vp.cos(angle_B), 210 * vp.sin(angle_B), 0)

# Vector C: 280  pointing 30 deg east of north (90-30)
angle_C = vp.radians(90 - 30)
C = vp.vec(280 * vp.cos(angle_C), 280 * vp.sin(angle_C), 0)

# --- 2. Calculate Resultant Vector R --
R = A + B + C

D = -R

# --- 3. Compute Magnitude and Angle ---
D_mag = D.mag

# Angle theta measured counterclockwise from the +x axis
theta_deg = vp.degrees(vp.atan2(D.y, D.x))
if theta_deg < 0:
    theta_deg += 360

# --- 4. Output Results ---

print(f"Vector D: {fmt_vec(D)} m")
print(f"Magnitude |D|: {D_mag:.1f} m")
print(f"Angle θ: {theta_deg:.0f}°")


print(f"Checking that she returns to the origin:")
print(f"Coordinates of vector A+B+C+D: {A + B + C + D}")
```

```python
# Ploting the vectors A, B, C, D
# Add root folder to sys.path
sys.path.append(str(Path("..").resolve()))

# Import your helper function directly
from helper_functions import plot_vector, plot_vector_xy

fig, ax = plt.subplots(figsize=(6, 6))
ax.set_aspect("equal")
ax.axhline(0, color="gray", linewidth=0.5)
ax.axvline(0, color="gray", linewidth=0.5)
ax.grid(True, linestyle="--", alpha=0.5)

# just in case is needed we can use ret, normally not used
# Vector A: 180 m at 180 deg from +y axis (straight west)
vec_a = plot_vector_xy(-180, 0, ax=ax, color="b", label="A (0, -180)")

# Vector B: 210 m pointing 45 deg east of south (270+45)
vec_b = plot_vector(
    210, 270 + 45, origin=((A.x, A.y)), ax=ax, color="g", label="B (210, 270+45°)"
)

# Vector C: 280  pointing 30 deg east of north (90-30)
vec_c = plot_vector(
    280,
    90 - 30,
    origin=((A + B).x, (A + B).y),
    ax=ax,
    color="y",
    label="C (280, 90-30°)",
)

# Vector D: 143.5 m pointing  221° (result from above solution)
vec_d = plot_vector(
    143.5,
    221,
    origin=((A + B + C).x, (A + B + C).y),
    ax=ax,
    color="r",
    label="D (143.5, 221°)",
)

plt.title("Vectors: A, B, C, D")

plt.show()

print(vec_a)
print(vec_b)
print(vec_b_minus_a)
```

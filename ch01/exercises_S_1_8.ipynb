The x-component of $\vec{A}$ is Ax = -20.0 m. 

**(a)** What is the y-component
of $\vec{A}$ 

```python
theta = vp.radians(90 - 40.0)
A_x = -20.0  # m

# tan(theta) = y / x ==> y = x tan(theta)
A_y = A_x * vp.tan(theta)

print(f"A_y = {A_y:.1f}")
```

(b) What is the magnitude of $\vec{A}$  ?

```python
A_mag = np.sqrt(A_x**2 + A_y**2)
print(f"Magnitude A: {A_mag:.1f} m")
```

#### 1.31 
•• A disoriented physics professor drives 3.25 km north, then
2.20 km west, and then 1.50 km south. Find the magnitude and direction
of the resultant displacement, using the method of components. In
a vector-addition diagram (roughly to scale), show that the resultant displacement
found from your diagram is in qualitative agreement with the
result you obtained by using the method of components.

```python
# Create vectors A, B, C for each path
A = vp.vec(0, 3.25, 0)  # km to the north
B = vp.vec(-2.20, 0, 0)  # km to the west
C = vp.vec(0, -1.50, 0)  # km to the south

# calculate displacement
D = A + B + C
print(f"vector D = {fmt_vec(D)}")
D_mag = D.mag
theta_D = vp.degrees(vp.atan2(D.y, D.x))
print(f"Magnitude D = {D_mag:.2f} km")
print(f"Angle theta = {theta_D % 360:.1f}°")
```

•• Vector $\vec{A}$ has magnitude 8.00 m and is in the xy-plane at an
angle of 127° counterclockwise from the +x–axis (37° past the +y-axis).
What are the magnitude and direction of vector $\vec{B}$ if the sum $\vec{A} + \vec{B}$ is
in the -y-direction and has magnitude 12.0 m?

```python
A_mag = 8.00  # m
A_angle = vp.radians(127)

A = vp.vec(A_mag * vp.cos(A_angle), A_mag * vp.sin(A_angle), 0)
C = vp.vec(0, 12.0, 0)  # C = A + B

# Compute B = C - A
B = C - A

# Calculate magnitude and direction
B_mag = B.mag

# Angle theta from x-axis
theta_deg = vp.degrees(vp.atan2(B.y, B.x))
theta_det = theta_deg % 360

print(f"Vector B: {fmt_vec(B)}")
print(f"Magnitude |B|: {B_mag:.2f} m")
print(f"Direction θ: {theta_deg:.1f}°")
```

•• Vector $\vec{A}$ is 2.80 cm long and is 60.0° above the x-axis in the
first quadrant. Vector $\vec{B}$ is 1.90 cm long and is 60.0° below the x-axis in
the fourth quadrant (Fig. E1.33). Use components to find the magnitude
and direction of 

**(a)** $\vec{A} + \vec{B}$ ; 


```python
A_mag = 2.80  # cm
A_theta = vp.radians(60.0)
B_mag = 1.90  # cm
B_theta = vp.radians(-60.0)

A = vp.vec(A_mag * vp.cos(A_theta), A_mag * vp.sin(A_theta), 0)
B = vp.vec(B_mag * vp.cos(B_theta), B_mag * vp.sin(B_theta), 0)

C = A + B
C_mag = C.mag
C_theta = vp.degrees(vp.atan2(C.y, C.x))

print(f"(a)")
print(f"C = A + B")
print(f"Vector C: {fmt_vec(C)}")
print(f"Magnitude |C|: {C_mag:.2f} cm")
print(f"Direction θ: {C_theta:.1f}°")
```


**(b)** $\vec{A} - \vec{B}$ ; 



```python
D = A - B
D_mag = D.mag
D_theta = vp.degrees(vp.atan2(D.y, D.x))

print(f"(b)")
print(f"D = A - B")
print(f"Vector D: {fmt_vec(D)}")
print(f"Magnitude |D|: {D_mag:.2f} cm")
print(f"Direction θ: {D_theta:.1f}°")
```

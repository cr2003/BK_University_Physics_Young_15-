• Given two vectors $\vec{A} = -2.00\hat{i} + 3.00\hat{j} + 4.00\hat{k}$
and $\vec{B} = 3.00\hat{i} + 1.00\hat{j} - 3.00\hat{k}$

**(a)** find the magnitude of each vector;


```python
A = vp.vec(-2.00, 3.00, 4.00)
B = vp.vec(3.00, 1.00, -3.00)

A_mag = A.mag
B_mag = B.mag

print(f"Magnitude |A| = {A_mag:.2f}")
print(f"Magnitude |B| = {B_mag:.2f}")
```

(b) use unit vectors to write an expression for the vector difference
$\vec{A} - \vec{B}$; 


```python
dif_A_B = A - B
```

and (c) find the magnitude of the vector difference $\vec{A} - \vec{B}$;  . Is
this the same as the magnitude of $\vec{B} - \vec{A}$;  ? Explain.

```python
dif_A_B_mag = dif_A_B.mag
dif_B_A = B - A
dif_B_A_mag = dif_B_A.mag

print(f"Magnitude |A - B| = {dif_A_B_mag:.2f}")
print(f"Magnitude |B - A| = {dif_B_A_mag:.2f}")
print("The magnitudes are the same. The vector B - A is simply the negative of A - B.")
print("Since calculating the magnitude involves squaring all component values,")
print("the negative sign cancels out.")
```

#### 1.49

•• White Dwarfs and Neutron Stars. 

Recall that density is mass
divided by volume, and consult Appendix D as needed. 

**(a)** Calculate the average
density of the earth in g/cm3, assuming our planet is a perfect sphere.




```python
# --- Earth data ---
R_earth = const.R_earth
M_earth = const.M_earth

# V = (4/3) pi r^3
V_earth = (4 / 3) * np.pi * R_earth**3
rho_earth = (M_earth / V_earth).to(u.g / u.cm**3)
print(f"Density of Earth = {rho_earth:.2f}")
```

**(b)** In about 5 billion years, at the end of its lifetime, our sun will end up as
a white dwarf that has about the same mass as it does now but is reduced
to about 15,000 km in diameter. What will be its density at that stage?


```python
# --- Sun data ---
R_sun = const.R_sun
M_sun = const.M_sun

# V = (4/3) pi r^3
V_sun = (4 / 3) * np.pi * R_sun**3
rho_sun = (M_sun / V_sun).to(u.g / u.cm**3)
print(f"Density of Sun = {rho_sun:.2f}")

# --- White Dwarf ---
R_wd = ((15_000 / 2) * u.km).to(u.m)
M_wd = M_sun

# V = (4/3) pi r^3
V_wd = (4 / 3) * np.pi * R_wd**3
rho_wd = (M_wd / V_wd).to(u.g / u.cm**3)
print(f"Density of White Dwarf = {rho_wd:.2e}")
```

**(c)** A neutron star is the remnant of certain supernovae (explosions of giant
stars). Typically, neutron stars are about 20 km in diameter and have about
the same mass as our sun. What is a typical neutron star density in g/cm3?

```python
# --- Neutron Star ---
R_ns = ((20 / 2) * u.km).to(u.m)
M_ns = M_sun

# V = (4/3) pi r^3
V_ns = (4 / 3) * np.pi * R_ns**3
rho_ns = (M_ns / V_ns).to(u.g / u.cm**3)
print(f"Density of Neutron Star = {rho_ns:.2e}")
```

#### 1.50 
 •• The Hydrogen Maser. 

A maser is a laser-type device that
produces electromagnetic waves with frequencies in the microwave
and radio-wave bands of the electromagnetic spectrum. You can use the
radio waves generated by a hydrogen maser as a standard of frequency.
The frequency of these waves is 1,420,405,751.786 hertz. (A hertz is another
name for one cycle per second.) A clock controlled by a hydrogen
maser is off by only 1 s in 100,000 years. For the following questions,
use only three significant figures. (The large number of significant figures
given for the frequency simply illustrates the remarkable accuracy
to which it has been measured.) 

**(a)** What is the time for one cycle of the
radio wave? 

```python
f = 1_420_405_751_786 * u.hertz

# Period
T = (1 / f).to(u.s)
print(f"(a) Period: {T.value:.2e} s")
```

**(b)** How many cycles occur in 1 h? 

```python
# (b) Cycles in 1 hour
# freq = cycles / time ==> cycles = freq * time
cycles_1h = (f * (1 * u.h)).decompose()

print(f"Cycles in 1h: {cycles_1h:.2e}")
```

**(c)** How many cycles
would have occurred during the age of the earth, which is estimated
to be $4.6 \times 10^9$ years?

```python
# (c) Cycles in age of Earth
t_earth = 4.6e9 * u.yr
cycles_earth = (f * t_earth).decompose()
print(f"(c) Cycles in Earth age: {cycles_earth.value:.2e}")
```

**(d)** By how many seconds would a hydrogen
maser clock be off after a time interval equal to the age of the earth?

```python
# (d) Clock off
error = (1 * u.s / (1e5 * u.yr)) * t_earth
print(f"(d) Clock error: {error.to(u.s).value:.2e} s ({error.to(u.h):.2f})")
```

In January 2006 astronomers reported
the discovery of a planet, comparable in size to the earth, orbiting another
star and having a mass about 5.5 times the earth’s mass. It is believed
to consist of a mixture of rock and ice, similar to Neptune. If this
planet has the same density as Neptune 1.76 g/cm3, what is its radius
expressed 

**(a)** in kilometers and 

**(b)** as a multiple of earth’s radius?
Consult the back of the book for astronomical data.

```python
# mass of the planet
m = 5.5 * const.M_earth
# density of the planet
rho = 1.76 * (u.g / u.cm**3)
# volume of the planet, similar to Earth
# rho = m / V ==> V = m / rho
V = m / rho


# V = (4 / 3) * np.pi * r**3 ==> r = cbrt(3V / 4 pi)
r = ((3 * V) / (4 * np.pi)) ** (1 / 3)
radius_km = r.to(u.km)

print(f"(a) Planet radius: {radius_km:,.2e}")
ratio = (r / const.R_earth).decompose()

print(f"(b) Planet radius: {ratio:,.2f} times Earth radius.")
```

#### 1.52 
••• A rectangular piece of aluminum is 7.60 $\pm$ 0.01 cm long
and 1.50 $\pm$ 0.01 cm wide. 

**(a)** Find the area of the rectangle and the
uncertainty in the area. 

**(b)** Verify that the fractional uncertainty in the
area is equal to the sum of the fractional uncertainties in the length and
in the width. (This is a general result.)

```python
import astropy.units as u

# Given dimensions
L, dL = 7.60 * u.cm, 0.01 * u.cm
W, dW = 1.50 * u.cm, 0.01 * u.cm

# (a) Area and Bounds
A = L * W
A_max = (L + dL) * (W + dW)
A_min = (L - dL) * (W - dW)
dA = (A_max - A_min) / 2

# (b) Fractional uncertainties
frac_L = (dL / L).decompose().value
frac_W = (dW / W).decompose().value
frac_A = (dA / A).decompose().value
sum_frac = frac_L + frac_W

print(f"(a) Area = {A.value:.2f} +/- {dA.value:.2f} {A.unit}")
print(f"(b) frac_A = {frac_A:.5f}")
print(f"    frac_L + frac_W = {sum_frac:.5f}")
```

#### 1.54 • 
BIO Biological tissues are typically made up of 98% water. Given
that the density of water is $1.0 \times 10^3 kg/m^3$, estimate the mass of 

**(a)** the heart of an adult human; 

**(b)** a cell with a diameter of $0.5 \mu m$; 

**(c)** a honeybee.

```python
# Density of tissue ~ density of water
rho_tissue = 1.0e3 * (u.kg / u.m**3)

# (a) Human Heart
# Approximating volume as 300 cm^3 (size of one fist)
V_heart = 300 * u.cm**3
m_heart = (rho_tissue * V_heart).to(u.kg)

# (b) Cell (diameter = 0.5 um)
r_cell = (0.5 / 2) * u.um
V_cell = (4 / 3) * np.pi * (r_cell**3)
m_cell = (rho_tissue * V_cell).to(u.kg)

# (c) Honeybee
# Approximating volume as ~0.1 cm^3 to 0.4 cm^3
V_bee = 0.1 * u.cm**3
m_bee = (rho_tissue * V_bee).to(u.g)

print(f"(a) Heart Mass:    {m_heart:.1f} ({m_heart.to(u.g):.0f})")
print(f"(b) Cell Mass:     {m_cell:.1e}")
print(f"(c) Honeybee Mass: {m_bee:.1f} ({m_bee.to(u.kg):.1e})")
```

•• Vector $\vec{A} = 3.0  - 4.0 \hat{k}$

**(a)** Construct a unit vector that
is parallel to $\vec{A}$. 

**(b)** Construct a unit vector that is antiparallel $\vec{A}$.

**(c)** Construct two unit vectors that are perpendicular to $\vec{A}$ and that have
no y-component.

```python
# 1. Define Vector A (3.0 * i - 4.0 * k)
A = vp.vector(3.0, 0.0, -4.0)

# Calculate magnitude
A_mag = vp.mag(A)

# --- Part (a): Parallel Unit Vector ---
# hat(v) returns the unit vector (v / mag(v))
u_parallel = vp.hat(A)

# --- Part (b): Antiparallel Unit Vector ---
u_antiparallel = -vp.hat(A)

# --- Part (c): Perpendicular Unit Vectors with no y-component ---
# Since A lies in the xz-plane, cross(A, vector(0, 1, 0)) gives a
# perpendicular direction in the xz-plane using the right-hand rule.
u_perp1 = vp.hat(vp.cross(A, vp.vec(0, 1, 0)))
u_perp2 = -u_perp1

# --- Display Results ---
print(f"Vector A: {A}")
print(f"Magnitude of A: {A_mag}\n")
print(f"(a) Parallel unit vector:     {u_parallel}")
print(f"(b) Antiparallel unit vector: {u_antiparallel}")
print(f"(c) Perpendicular unit 1:     {u_perp1}")
print(f"    Perpendicular unit 2:     {u_perp2}")
```

ground, producing the vector forces $\vec{A}$ , $\vec{B}$, and $\vec{C}$ shown in Fig. P1.56.
Find the magnitude and direction of a fourth force on the stone that will
make the vector sum of the four forces zero.

```python
mag_A = 100.0  # N
angle_A = vp.radians(30.0)
mag_B = 80.0  # N
angle_B = vp.radians(90 + 30.0)
mag_C = 40.0  # N
angle_C = vp.radians(180 + 53.0)

A = vp.vec(mag_A * vp.cos(angle_A), mag_A * vp.sin(angle_A), 0)
B = vp.vec(mag_B * vp.cos(angle_B), mag_B * vp.sin(angle_B), 0)
C = vp.vec(mag_C * vp.cos(angle_C), mag_C * vp.sin(angle_C), 0)
R = A + B + C

# calculate the 4 force ==> D
D = -R
mag_D = D.mag
angle_D = vp.degrees(vp.atan2(D.y, D.x)) % 360

print(f"Magnitude |D| = {mag_D:.1f}, theta={angle_D:.1f}")

# check the sum of 4 vetors is null
print(f"Sum A + B + C + D = {A + B + C + D}")
```

•• A sailor in a small sailboat encounters shifting winds. She
sails 2.00 km east, next 3.50 km southeast, and then an additional distance
in an unknown direction. Her final position is 5.80 km directly
east of the starting point (Fig. P1.60). Find the magnitude and direction
of the third leg of the journey. Draw the vector-addition diagram and
show that it is in qualitative agreement with your numerical solution.

```python
# Legs of the journey
l_1 = vp.vec(2, 0, 0)  # km
l_2_mag = 3.50  # km
l_2_angle = vp.radians(360 - 45.0)
l_2 = vp.vec(l_2_mag * vp.cos(l_2_angle), l_2_mag * vp.sin(l_2_angle), 0)

# Displacement d
d = vp.vec(5.80, 0, 0)  # km


# d = l_1 + l_2 + l_3 ==> l_3 = d - l_1 - l_2
l_3 = d - l_1 - l_2

# calculate magnitud and direction of l_3
l_3_mag = l_3.mag
l_3_angle = vp.degrees(vp.atan2(l_3.y, l_3.x)) % 360

print(f"Vector l_3 = <{l_3_mag:.0f} km, {l_3_angle:.1f}°>")

# Check the result
print("\nCheck the result:")
print(f"l_1 + l_2 + l_3 = {l_1 + l_2 + l_3}")
```

```python
fig, ax = plt.subplots(figsize=(6, 6))
ax.set_aspect("equal")
ax.axhline(0, color="gray", linewidth=0.5)
ax.axvline(0, color="gray", linewidth=0.5)
ax.grid(True, linestyle="--", alpha=0.5)

# Plot vectors
vec_l1 = plot_vector_xy(2.00, 0, ax=ax, color="b", label="(vector l1 (2.00 km, 0°))")
vec_l2 = plot_vector(
    3.50,
    360 - 45.0,
    origin=(l_1.x, l_1.y),
    ax=ax,
    color="b",
    label="(vector l2 (3.50, 360 - 45.0°))",
)
vec_l3 = plot_vector(
    l_3.mag,
    l_3_angle,
    origin=((l_1 + l_2).x, (l_1 + l_2).y),
    ax=ax,
    color="b",
    label="(vector l3 (3.00 km, 61.8°))",
)
# --- Add a red asterisk at point (x=0, y=0) ---
ax.plot(0, 0, "r*", markersize=12, label="Start")
ax.plot(5.8, 0, "r*", markersize=12, label="Finish")


plt.title("Vectors: Problem 1.60")


plt.tight_layout()
plt.show()
```

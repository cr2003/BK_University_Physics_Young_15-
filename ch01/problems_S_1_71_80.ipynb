•• The work W done by a constant force $\vec{F}$ on an object that undergoes
displacement $\vec{s}$ from point 1 to point 2 is $W = \vec{F} \cdot \vec{s}$.
For F in newtons (N) and s in meters (m), W is in joules (J). If, during a displacement
of the object, $\vec{F}$ has constant direction 60.0° above the -x-axis
and constant magnitude 5.00 N and if the displacement is 0.800 m in the
+x-direction, what is the work done by the force $\vec{F}$?

```python
theta = np.radians(180 - 60)
F_mag = 5  # N
F = vp.vec(5 * np.cos(theta), 5 * np.sin(theta), 0)
s = vp.vec(0.800, 0, 0)  # m
W = vp.dot(F, s)
print(f"Work W = {W:.2f} J")
```

•• Magnetic fields are produced by moving charges and exert
forces on moving charges. When a particle with charge q is moving
with velocity vS in a magnetic field $\vec{B}$ , the force $\vec{F}$ that the field exerts
on the particle is given by $\vec{F} = q\vec{v}\times \vec{B}$. The SI units are as follows: For
charge it is the coulomb (C), for magnetic field it is tesla (T), for force
it is newton (N), and for velocity it is m/s. If $q = -8.00 \times 10^6 C$, 
$\vec{v}$ is $3.00 \times 10^4 m/s$ in the +x-direction, and $\vec{B}$ is 5.00 T in the -y-direction,
what are the magnitude and direction of the force that the magnetic
field exerts on the charged particle?

```python
# --- vPython Evaluation Block ---
q_val = -8.00e-6 * u.C
vx_val = 3.00e4 * (u.m / u.s)
By_val = 5.00 * u.T

v_vec = vp.vec(vx_val.value, 0, 0)  # (u.m / u.s)
B_vec = vp.vec(0, -By_val.value, 0)  #  u.T

# Calculate Force vector
# F = q * (v x B)
v_cross_B = vp.cross(v_vec, B_vec)  # (u.m / u.s * u.T)
F_vec = q_val.value * v_cross_B  # (u.N)

F_mag = F_vec.mag  # u.N

print(f"Velocity Vector v:       {v_vec} m/s")
print(f"Magnetic Field Vector B: {B_vec} T")
print(f"Force Vector F:          {F_vec} N")
print(f"Force Magnitude |F|:     {F_mag:.2f} N")
```

•• Torque is a vector quantity that specifies the effectiveness of
a force in causing the rotation of an object. The torque that a force $\vec{F}$
exerts on a rigid object depends on the point where the force acts and on
the location of the axis of rotation. If $\vec{r}$ is the length vector from the axis
to the point of application of the force, then the torque is $\vec{r} \times \vec{F}$.
If $\vec{F}$ is 22.0 N in the -y-direction and if $\vec{r}$ is in the xy-plane at an angle of 36°
from the +y-axis toward the -x-axis and has magnitude 4.0 m, what are
the magnitude and direction of the torque exerted by $\vec{F}$?

```python
F = vp.vec(0, -22, 0)  # N
r_mag = 4.0  # m
theta = vp.radians(90 + 36)
r = vp.vec(r_mag * vp.cos(theta), r_mag * vp.sin(theta), 0)

# calculate the torque (tau)
tau = vp.cross(r, F)

print(f"Position Vector r: {fmt_vec(r, 1)} m")
print(f"Force Vector F:    {fmt_vec(F, 1)} N")
print(f"Torque Vector tau: {fmt_vec(tau, 1)} N m")
print(f"Torque Magnitude:  {tau.mag:.1f} N m")
print(f"Torque direction:  {tau.hat} ")
```

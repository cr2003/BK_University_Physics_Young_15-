**BIO Calculating Lung Volume in Humans**. In humans, oxygen and
carbon dioxide are exchanged in the blood within many small sacs
called alveoli in the lungs. Alveoli provide a large surface area for gas
exchange. Recent careful measurements show that the total number of
alveoli in a typical pair of lungs is about $480 \times 10^6$ and that the average
volume of a single alveolus is $4.2 \times 10^6 \mu m^3$. (The volume of a sphere
is $V = \frac{4}{3}\pi r^3$, and the area of a sphere is $A = 4\pi r^2$.)


1.92

page 61

What is total volume of the gas-exchanging region of the lungs?

```python
N = 480e6  # Total number of alveoli
V_alveolus = 4.2e6 * u.um**3  # Volume of a single alveolus
V_total = N * V_alveolus
print(f"total volume: {V_total.to(u.l):.0f}")
```

If we assume that alveoli are spherical, what is the diameter of a
typical alveolus? 

```python
# V = (4/3) pi r^3 ==> r = (3V / 4 pi)**(1/3)
r = ((3 * V_alveolus) / (4 * np.pi)) ** (1 / 3)
d = 2 * r
print(f"Diameter: {d.to(u.mm):.2f}")
```

### EXAMPLE 2.4 Constant-acceleration calculations

page 76

<img src="..//pics/Fig_2_20.png" width=500> 

A motorcyclist heading east through a small town accelerates at a
constant $4.0 m/s^2$ after he leaves the city limits (Fig. 2.20). At time
t = 0 he is 5.0 m east of the city-limits signpost while he moves east at
15 m/s. 

**(a)** Find his position and velocity at t = 2.0 s. 




```python
a_x = 4.0 * (u.m / u.s**2)
t0 = 0 * u.s
x_0 = 5.0 * u.m
v_0x = 15 * (u.m / u.s)
t = 2 * u.s

# --- (a) Find his position and velocity at t = 2.0 s ---
x = x_0 + v_0x * t + (1 / 2) * a_x * t**2
v = v_0x + a_x * t

print("(a)")
print("At t = 2.0 s")
print(f"Position x = {x}")
print(f"Velocity v = {v}")

# --- (b)  Where is he when his speed is 25 m/s? ---
# v^2 = v_0^2 + 2 a (x - x0) ==> x = x0 + v^2 - V0^2 / 2 a
v = 25 * (u.m / u.s)

x = x_0 + (v**2 - v_0x**2) / (2 * a_x)
```

**(b)** Where is he
when his speed is 25 m/s?

```python
print("(b)")
print("when his speed is 25 m/s")
print(f"Position x = {x}")
```

### EXAMPLE 2.5 Two objects with different accelerations

page 76

<img src="..//pics/Fig_2_21.png" width=800> 

A motorist traveling at a constant 15 m/s (54 km/h) passes a school
crossing where the speed limit is 10 m/s (36 km/h). Just as the motorist
passes the school-crossing sign, a police officer on a motorcycle stopped
there starts in pursuit with constant acceleration 3.0 m/s2 (Fig. 2.21a).


**(a)** How much time elapses before the officer passes the motorist? At
that time, 




```python
### a

# Given values
vM0 = 15.0 * (u.m / u.s)
aP = 3.0 * (u.m / (u.s**2))

# xp = xp0 + vp0 t + (1/2)a p t^2 = (1/2)ap t^2 # eq. position police
# xM = vM0 t                                    # e1. position motorist
# we set xp = xM and solve for t
# vM0 t = (1/2)ap t^2 ==> t = (2 vM0) /  ap
t = (2 * vM0) / aP
print(f"The officer catches the motorist at time t: {t}")
```


**(b)** what is the officer’s speed and 


```python
### b
# vP = v0 + aP t # v0 = 0 (the officer was at rest)
vP = aP * t
vP
print(f"(b) Officer's speed: {vP:.1f}")
```


**(c)** how far has each vehicle
traveled?

```python
### c
# in 10 s the motorist travels a distance
xM = vM0 * t
xM

# in 10 s the officer travels a distance
xp = (1 / 2) * aP * t**2
xp
print(f"(c) Distance traveled by motorist: {xM:.1f}")
print(f"    Distance traveled by officer: {xp:.1f}")
```

### EXAMPLE 2.7Up-and-down motion in free fall

page 79

<img src="..//pics/Fig_2_24.png" width=400> 

You throw a ball vertically upward from the roof of a tall building.
The ball leaves your hand at a point even with the roof railing with
an upward speed of 15.0 m/s; the ball is then in free fall. (We ignore
air resistance.) On its way back down, it just misses the railing. Find

**(a)** the ball’s position and velocity 1.00 s and 4.00 s after leaving your
hand; 



```python
v0y = 15 * (u.m / u.s)
g = 9.80 * (u.m / u.s**2)
ay = -g
t0 = 0 * u.s
t1 = 1 * u.s
t4 = 4 * u.s
y0 = 0 * u.m  # origin at roof rail

# eq. position and velocity
# y = y0 + v0y * t + (1/2) *ay * t**2
# vy = v0y + ay * t

# at t1 = 1s
y = y0 + v0y * t1 + (1 / 2) * ay * t1**2
vy = v0y + ay * t1
print(f"At time t={t1}, y={y:.1f}, vy={vy:.1f}")

# at t4 = 4s
y = y0 + v0y * t4 + (1 / 2) * ay * t4**2
vy = v0y + ay * t4
print(f"At time t={t4}, y={y:.1f}, vy={vy:.1f}")
```

**(b)** the ball’s velocity when it is 5.00 m above the railing; 

```python
# The y-velocity at any position y is given by Eq.
# vy^2 = v0y^2 + 2ay (y-y0) ==> vy = ( v0y^2 + 2ay (y-y0))**(1/2)
# When the ball is 5.00 m above the origin we have y = +5.00 m
y = 5 * u.m
vy = (v0y**2 + 2 * ay * (y - y0)) ** (1 / 2)
print(f"at y={y}, vy = {vy:.1f}")
print("Note that we have a square root in the equation, so we have two solutions:")
print(f"+{vy:.1f} the way up and -{vy:.1f} the way down.")
```

**(c)** the maximum height reached; 

```python
# At the apex y_max, vy = 0
# with vy=0, y0=0, and y_max=0 the eq
# vy^2 = v0y^2 + 2ay (y-y0) ==> 0 = v0y^2 + (y_max -0) ==> y_max = -v0y^2 /(2a)
y_max = -(v0y**2) / (2 * ay)
print(f"the maximum height reached y_max={y_max:.1f}")
```

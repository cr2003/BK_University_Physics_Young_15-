#### VP2.5.1 

A sports car starts from rest at an intersection and accelerates
toward the east on a straight road at 8.0 m/s2. Just as the sports
car starts to move, a bus traveling east at a constant 18 m/s on the same
straight road passes the sports car. When the sports car catches up with
and passes the bus, 

**(a)** how much time has elapsed and 



```python
### a

# Given values
a_s = 8.0 * (u.m / (u.s**2))  # acceleration sports car
v_0b = 18.0 * (u.m / u.s)  # velocity of Bus


# x_s = x_s0 + vs0 t + (1/2)a_s p t^2 = (1/2)a_s t^2 # eq. position sports car,
# x_b = v_0b t                                    # e1. position bus
# we set x_s = x_b and solve for t
# v_0b t  = (1/2)a_s t^2 ==> t = (2 v_0b) /  a_s
t = (2 * v_0b) / a_s
print(f"The sports car catches the bus at time t: {t}")
```

(**b)** how far has the sports car traveled?

```python
# in 4.5 s the sports car travels a distance
x_s = (1 / 2) * a_s * t**2

print(f"Distance traveled by the sports car: {x_s:.1f}")
```

#### Exercise 2.1

• A car travels in the +x-direction on a straight and level road.
For the first 4.00 s of its motion, the average velocity of the car is
$v_{av-x} = 6.25 m/s$. How far does the car travel in 4.00 s?

```python
t = 4.00 * u.s
v_av = 6.25 * (u.m / u.s)

x = v_av * t
print(f"car travel: x = {x:.1f} in the x-direction")
```

#### 2.3 
•• Trip Home. You normally drive from Glasgow to Edinburgh
via M8 motorway at an average speed of 72 km/h, and the trip takes 1 h
and 2 min. At peak times, however, heavy traffic slows you down and
you drive the same distance at an average speed of only 49 km/h. How
much longer does the trip take?

```python
v_avg1 = 72 * (u.km / u.h)
t_1 = 1 * u.h + 2 * u.min

v_avg2 = 49 * (u.km / u.h)

# v_avg = delta_x / delta_t ==> delta_x = v_avg * delta_t

delta_x = v_avg1 * t_1

# delta_t = delta_x / v_avg
t_2 = delta_x / v_avg2

delta_t = (t_2 - t_1).to(u.min)
delta_t
print(f"The trip take {delta_t:.1f} equivalent to {delta_t.to(u.s):,.0f} longer")
```

#### 2.4 
•• From Pillar to Post. Starting from a pillar, you run 200 m east
(the +x-direction) at an average speed of 5.0 m/s and then run 280 m
west at an average speed of 4.0 m/s to a post. 

Calculate 

**(a)** your average speed from pillar to post and 



```python
# Calculate time for each leg
# Leg 1 (East)
d1 = 200 * u.m
v1 = 5.0 * (u.m / u.s)
t1 = d1 / v1
print(f"t1 = {t1}")

# Leg 2 (West)
d2 = 280 * u.m  # - because west direction is in the -x-direction
v2 = 4.0 * (u.m / u.s)
t2 = d2 / v2
print(f"t2 = {t2}")

delta_t = t1 + t2
print(f"total time: delta_t = {delta_t}")

# Average speed from pillar to post
# Average speed is the total distance traveled divided by the total time taken
s_avg = (d1 + d2) / delta_t
print(f"Average speed: s_avg = {s_avg:.2f}")
```

**(b)** your average velocity from pillar to post.

```python
# Average velocity from pillar to post
# Average velocity is the total displacement divided by the total time taken
delta_x = d1 - d2
delta_t = t1 + t2
print(f"Displacement: delta_x = {delta_x}")
v_avg = delta_x / delta_t
print(f"Average velocity: v_avg = {v_avg:.2f} due west")
```

#### 2.6 
•• A Honda Civic travels in a straight line along a road. The
car’s distance x from a stop sign is given as a function of time
t by the equation $x(t) = \alpha t^2 - \beta t^3$ , where $\alpha = 1.50 m/s^2$ and
$\beta = 0.0500 m/s^3$. Calculate the average velocity of the car for
each time interval: 

**(a)** t = 0 to t = 2.00 s; 


```python
# define the function x(t)
def x(t):
    alpha = 1.50 * (u.m / u.s**2)
    beta = 0.0500 * (u.m / u.s**3)
    return alpha * t**2 - beta * t**3


# v_avg = x(t2) - x(t1) / t2 - t1
# Calculate x(t) for different times
t0 = 0 * u.s
t2 = 2.00 * u.s
x0 = x(t0)
x2 = x(t2)

# (a) time interval t=0 to t=2.00s
v_avg_a = (x2 - x0) / (t2 - t0)
v_avg_a
print(f"For time interval (a) the average velocity: v_avg_a = {v_avg_a:.2f}")
```

**(b)** t = 0 to t = 4.00 s;


```python
# (b) time interval t=0 to t=4.00s
t4 = 4.00 * u.s
x4 = x(t4)
v_avg_b = (x4 - x0) / (t4 - t0)
v_avg_b
print(f"For time interval (b) the average velocity: v_avg_b = {v_avg_b:.2f}")
```

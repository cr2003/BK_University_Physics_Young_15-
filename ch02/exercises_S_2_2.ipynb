#### 2.7 

• CALC A car is stopped at a traffic light. It then travels
along a straight road such that its distance from the light is given
by $x(t) = bt^2 -ct^3$, where b = 2.40 m/s2 and c = 0.120 m/s3.

**(a)** Calculate the average velocity of the car for the time interval t = 0
to t = 10.0 s. 


```python
# define the position function x(t)
def x(t):
    b = 2.40 * (u.m / u.s**2)
    c = 0.120 * (u.m / u.s**3)
    return b * t**2 - c * t**3


# v_avg = x(t2) - x(t1) / t2 - t1
# Calculate x(t) for different times
t0 = 0 * u.s
t10 = 10.0 * u.s
x0 = x(t0)
x10 = x(t10)

# (a) time interval t=0 to t=10.0s
v_avg_a = (x10 - x0) / (t10 - t0)

print(f"For time interval (a) the average velocity: v_avg_a = {v_avg_a:.1f}")
```

**(b)** Calculate the instantaneous velocity of the car at
t = 0, t = 5.0 s, and t = 10.0 s. 

```python
# Define the symbolic time variable and constants
t = sp.Symbol("t")
b = sp.Symbol("b")
c = sp.Symbol("c")

# Define position function x(t)
x = b * t**2 - c * t**3

# Compute velocity v(t) as the derivative of x(t) with respect to t
v = sp.diff(x, t)

# Display the resulting expression
display(Math(rf"v = {sp.latex(v)}"))
```

```python
def v(t):
    b = 2.40 * (u.m / u.s**2)
    c = 0.120 * (u.m / u.s**3)
    return 2 * b * t - 3 * c * t**2


v0 = v(t0)
print(f"At t = {t0} instantaneous velocity {v0}")

t5 = 5.0 * u.s
v5 = v(t5)
print(f"At t = {t5} instantaneous velocity {v5}")

t10 = 10.0 * u.s
v10 = v(t10)
print(f"At t = {t10} instantaneous velocity {v10}")
```


(c) How long after starting from rest
is the car again at rest?

```python
# the car is at rest when v(t) = 0, so at t=0 and solving for t
# v(t) = 2bt - 3ct^2 ==> 0 = 2bt - 3ct^2 ==> t = (2b)/(3c)
b = 2.40 * (u.m / u.s**2)
c = 0.120 * (u.m / u.s**3)
t = 2 * b / (3 * c)
print(f"The car is again at rest at t = {t:.2f}")
```

given by  $x(t) = 28.0 m + (12.4 m/s)t - (0.0450 m/s^3)t^3$. What is the
instantaneous velocity of the bird when t = 8.00 s?

```python
# Define the symbolic time variable and constants
a, b, c, t = sp.symbols("a, b, c, t")

# Define position function x(t)
x = a + b * t - c * t**3

# Compute velocity v(t) as the derivative of x(t) with respect to t
v = sp.diff(x, t)

# Display the resulting expression
display(Math(rf"v(t) = {sp.latex(v)}"))
```

•• A ball moves in a straight
line (the x-axis). The graph in
Fig. E2.9 shows this ball’s velocity
as a function of time. 

**(a)** What are
the ball’s average speed and average
velocity during the first 3.0 s?


```python
# We calculate displacement (delta_x)
# Leg 1 (East) from t=0 to t=2.0 s:
delta_t1 = (2 - 0) * u.s
v1 = 2.0 * (u.m / u.s)  # taken from the graph
# v = delta_x / delta_t ==> delta_x = v * delta_t
delta_x1 = v1 * delta_t1

# Leg 2 (East) from t=2 to t=3
delta_t2 = (3 - 2) * u.s
v2 = 3.0 * (u.m / u.s)  # taken from the graph
delta_x2 = v2 * delta_t2

# Total displacement and total distance
delta_x = delta_x1 + delta_x2
d = np.abs(delta_x1) + np.abs(delta_x2)

# total time
delta_t = delta_t1 + delta_t2

# average speed
s_avg = d / delta_t

# average velocity
v_avg = delta_x / delta_t

# Answer (a)
print(f"Average speed s_avg={s_avg:.2f}")
print(f"Average velocity v_avg={v_avg:.2f}")
```

(b) Suppose that the ball moved in
such a way that the graph segment
after 2.0 s was -3.0 m>s instead of
+3.0 m>s. Find the ball’s average
speed and average velocity in this
case.

```python
# We calculate displacement (delta_x)
# Leg 1 (East) from t=0 to t=2.0 s:
delta_t1 = (2 - 0) * u.s
v1 = 2.0 * (u.m / u.s)  # taken from the graph
# v = delta_x / delta_t ==> delta_x = v * delta_t
delta_x1 = v1 * delta_t1

# Leg 2 (West) from t=2 to t=3
delta_t2 = (3 - 2) * u.s
v2 = -3.0 * (u.m / u.s)  # taken from the graph
delta_x2 = v2 * delta_t2

# Total displacement and total distance
delta_x = delta_x1 + delta_x2
d = np.abs(delta_x1) + np.abs(delta_x2)

# total time
delta_t = delta_t1 + delta_t2

# average speed
s_avg = d / delta_t

# average velocity
v_avg = delta_x / delta_t

# Answer (a)
print(f"Average speed s_avg={s_avg:.2f}")
print(f"Average velocity v_avg={v_avg:.2f}")
```

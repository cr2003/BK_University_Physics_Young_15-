#### 2.49 
• **CALC** A rocket starts from rest and moves upward from the
surface of the earth. For the first 10.0 s of its motion, the vertical acceleration
of the rocket is given by $a_y = (2.80 m/s^3)t$, where the
+y-direction is upward. 

**(a)** What is the height of the rocket above the
surface of the earth at t = 10.0 s? 

**(b)** What is the speed of the rocket
when it is 325 m above the surface of the earth?

```python
# Define symbolic variables
t, C, y = sp.symbols("t C y", real=True, positive=True)

# Define acceleration and integrate
a_y = C * t
v_expr = sp.integrate(a_y, t)  # v(t) = C*t^2 / 2
y_expr = sp.integrate(v_expr, t)  # y(t) = C*t^3 / 6

# Express v as a function of y directly by substituting t(y) = (6*y/C)^(1/3)
t_from_y = (6 * y / C) ** sp.Rational(1, 3)
v_from_y_expr = sp.simplify(v_expr.subs(t, t_from_y))

display(Math(rf"v_y(t) = {sp.latex(v_expr)}"))
display(Math(rf"y(t) = {sp.latex(y_expr)}"))
display(Math(rf"v_y(y) = {sp.latex(v_from_y_expr)}"))
```

```python
# Create lambdified functions using symbol tuples
calc_y = sp.lambdify((t, C), y_expr)
calc_v = sp.lambdify((y, C), v_from_y_expr)  # Uses v_from_y_expr, which depends on y

# --- Physical Inputs ---
C_qty = 2.80 * (u.m / (u.s**3))
t_qty = 10.0 * u.s
y_qty = 325.0 * u.m

# --- Calculations ---
height_a = calc_y(t_qty, C_qty)
speed_b = calc_v(y_qty, C_qty)

print(f"Part (a) Height at t = 10.0 s : {height_a:.3g}")
print(f"Part (b) Speed at y = 325 m   : {speed_b:.3g}")
```

#### 2.51 

•• **CALC** The acceleration of a motorcycle is given by
$a_x(t) = At - Bt2$, where A = 1.50 m/s3 and B = 0.120 m/s4. The
motor cycle is at rest at the origin at time t = 0. 

**(a)** Find its position and velocity as functions of time. 

**(b)** Calculate the maximum velocity it attains.

```python
# Define symbolic variables
t, A, B = sp.symbols("t A B", real=True, positive=True)

# Acceleration definition
a_x = A * t - B * t**2

# Integrate for velocity and position (initial conditions = 0)
v_expr = sp.integrate(a_x, t)
x_expr = sp.integrate(v_expr, t)

# Critical time when acceleration is zero
t_max_expr = sp.solve(sp.Eq(a_x, 0), t)[0]  # Select t = A/B

# Maximum velocity formula
v_max_expr = sp.simplify(v_expr.subs(t, t_max_expr))

display(Math(rf"v_x(t) = {sp.latex(v_expr)}"))
display(Math(rf"x(t) = {sp.latex(x_expr)}"))
display(Math(rf"t_{{max}} = {sp.latex(t_max_expr)}"))
display(Math(rf"v_{{max}} = {sp.latex(v_max_expr)}"))
```

```python
# --- 1. Symbolic Variables & Derivations ---
t_s, A_s, B_s = sp.symbols("t A B", real=True, positive=True)

a_expr = A_s * t_s - B_s * t_s**2
v_expr = sp.integrate(a_expr, t_s)
x_expr = sp.integrate(v_expr, t_s)

# --- 2. Numerical Values ---
A_val = 1.50
B_val = 0.120

# --- 3. Substitute A and B to get pure numeric formulas ---
v_numeric_expr = v_expr.subs({A_s: A_val, B_s: B_val})
x_numeric_expr = x_expr.subs({A_s: A_val, B_s: B_val})

# Print formulas with numeric coefficients
print("v_x(t) =", v_numeric_expr)
# Output: v_x(t) = -0.04*t**3 + 0.75*t**2

print("x(t)   =", x_numeric_expr)
# Output: x(t)   = -0.01*t**4 + 0.25*t**3

# --- 4. Evaluate using lambdify with the substituted formulas ---
calc_v_num = sp.lambdify(t_s, v_numeric_expr)
calc_x_num = sp.lambdify(t_s, x_numeric_expr)

# Evaluate at a specific time (e.g., t = 12.5 s)
t_qty = 12.5 * u.s

v_at_t = calc_v_num(t_qty.value) * (u.m / u.s)
x_at_t = calc_x_num(t_qty.value) * u.m

print(f"At t = {t_qty}: v_x = {v_at_t:.2f}, x = {x_at_t:.2f}")
```

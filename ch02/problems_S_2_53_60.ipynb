#### 2.53

• **BIO** A typical male sprinter can maintain his maximum acceleration
for 2.0 s, and his maximum speed is 10 m/s. After he reaches
this maximum speed, his acceleration becomes zero, and then he runs at
constant speed. Assume that his acceleration is constant during the first
2.0 s of the race, that he starts from rest, and that he runs in a straight
line. 

**(a)** How far has the sprinter run when he reaches his maximum
speed? 

```python
# --- Given parameters ---
t_qty = 2.0 * u.s
v_qty = 10 * (u.m / u.s)
v0_qty = 0 * (u.m / u.s)

# From the equation a = (v - v0) / t
a_qty = (v_qty - v0_qty) / t_qty

# From the equation x = x0 + v0 t + 1/2 a t^2 (When x0=0, v0=0) ==> x = 1/2 a t^2
x_qty = (1 / 2) * a_qty * t_qty**2

print("--- Part (a) ---")
print(f"Calculated acceleration a = {a_qty:.2g}")
print(f"At time {t_qty} the sprinter has run m = {x_qty:.2g} ")
```

**(b)** What is the magnitude of his average velocity for a race of
these lengths: 
* (i) 50.0 m; 

* (ii) 100.0 m; 

* (iii) 200.0 m?

```python
# As the sprinter needs 2 seconds for the first 10 m, and then keep his speed at 10 m/s
# time_race = (len_race / 10) + 1
def avg_speed(len_race):
    time_race = (len_race.value / 10 + 1) * u.s
    speed = len_race / time_race
    return speed


len_race1_qty = 50.0 * u.m
len_race2_qty = 100.0 * u.m
len_race3_qty = 200.0 * u.m

avg_s1_qty = avg_speed(len_race1_qty)
avg_s2_qty = avg_speed(len_race2_qty)
avg_s3_qty = avg_speed(len_race3_qty)

print("--- Part (b) ---")
print(
    f"for a race of {len_race1_qty} the magnitude of average speed = {avg_s1_qty:.3g}"
)
print(
    f"for a race of {len_race2_qty} the magnitude of average speed = {avg_s2_qty:.3g}"
)
print(
    f"for a race of {len_race3_qty} the magnitude of average speed = {avg_s3_qty:.3g}"
)
```

#### 2.55 

••• **Earthquake Analysis**. Earthquakes produce several types
of shock waves. The most well known are the P-waves (P for primary or
pressure) and the S-waves (S for secondary or shear). In the earth’s crust,
P-waves travel at about 6.5 km/s and S-waves move at about 3.5 km/s.
The time delay between the arrival of these two waves at a seismic recording
station tells geologists how far away an earthquake occurred. If the time
delay is 33 s, how far from the seismic station did the earthquake occur?

```python
# Define symbolic variables
d, v_P, v_S, delta_t = sp.symbols("d v_P v_S Delta_t", real=True, positive=True)

# Travel times
t_P = d / v_P
t_S = d / v_S

# Time delay equation
delay_eq = sp.Eq(delta_t, t_S - t_P)

# Solve for distance d
d_expr = sp.solve(delay_eq, d)[0]
d_expr_simplified = sp.simplify(d_expr)

display(Math(rf"\Delta t = {sp.latex(t_S - t_P)}"))
display(Math(rf"d = {sp.latex(d_expr_simplified)}"))
```

```python
# --- 3. Astropy Execution with Units ---
vP_qty = 6.5 * (u.km / u.s)
vS_qty = 3.5 * (u.km / u.s)
dt_qty = 33.0 * u.s

calc_d = sp.lambdify((v_P, v_S, delta_t), d_expr)
d_qty = calc_d(vP_qty, vS_qty, dt_qty)

print(f"\nDistance to epicenter : {d_qty:.2f}")
print(f"Distance (2 sig figs) : {d_qty:.0f}")
```

#### 2.58 

•• A block moving on a horizontal surface is at x = 0 when
t = 0 and is sliding east with a speed of 12.0 m/s. Because of a net force
acting on the block, it has a constant acceleration with direction west
and magnitude 2.00 m/s2. The block travels east, slows down, reverses
direction, and then travels west with increasing speed. 

**(a)** At what value
of t is the block again at x = 0? 

**(b)** What is the maximum distance
east of x = 0 that the rock reaches, and how long does it take the
rock to reach this point?

```python
# Define symbolic variables
t, v0, a = sp.symbols("t v_0 a", real=True)

# Kinematic equations
v_expr = v0 + a * t
x_expr = v0 * t + sp.Rational(1, 2) * a * t**2

# (a) Time to return to x = 0 (t > 0 root)
t_return_expr = sp.solve(sp.Eq(x_expr, 0), t)[1]

# (b) Time when max distance is reached (v = 0)
t_max_expr = sp.solve(sp.Eq(v_expr, 0), t)[0]

# Maximum distance east
x_max_expr = sp.simplify(x_expr.subs(t, t_max_expr))

display(Math(rf"x(t) = {sp.latex(x_expr)}"))
display(Math(rf"t_{{return}} = {sp.latex(t_return_expr)}"))
display(Math(rf"t_{{max}} = {sp.latex(t_max_expr)}"))
display(Math(rf"x_{{max}} = {sp.latex(x_max_expr)}"))
```

```python
# --- 1. Create Lambdified Functions from Cell 1 Expressions ---
calc_t_return = sp.lambdify((v0, a), t_return_expr)
calc_t_max = sp.lambdify((v0, a), t_max_expr)
calc_x_max = sp.lambdify((v0, a), x_max_expr)

# --- 2. Astropy Physical Inputs ---
v0_qty = 12.0 * (u.m / u.s)
a_qty = -2.00 * (u.m / (u.s**2))

# --- 4. Evaluate Lambdified Functions ---
t_ret_qty = calc_t_return(v0_qty.value, a_qty.value) * u.s
t_peak_qty = calc_t_max(v0_qty.value, a_qty.value) * u.s
x_peak_qty = calc_x_max(v0_qty.value, a_qty.value) * u.m

print(f"Part (a) Return time to x = 0 : {t_ret_qty:.2f}")
print(f"Part (b) Time to reach peak  : {t_peak_qty:.2f}")
print(f"Part (b) Max distance east   : {x_peak_qty:.2f}")
```

#### 2.59 

•• A block is sliding with constant acceleration down an incline.
The block starts from rest at t = 0 and has speed 3.00 m/s after it has
traveled a distance 8.00 m from its starting point. 

**(a)** What is the speed of
the block when it is a distance of 16.0 m from its t = 0 starting point? 

**(b)** How long does it take the block to slide 16.0 m from its starting point?

```python
# Define symbolic variables
v, v0, a, x, t = sp.symbols("v, v_0, a, x, t", real=True)

# Kinematic equations
v_expr_sq = v0**2 + 2 * a * x
x_expr = v0 * t + sp.Rational(1, 2) * a * t**2

# Acceleration
a_expr = sp.solve(sp.Eq(v**2, v_expr_sq), a)[0]
v_expr = sp.solve(sp.Eq(v**2, v_expr_sq), v)[1]
t_expr = sp.solve(sp.Eq(x, x_expr), t)[1]

display(Math(rf"a = {sp.latex(a_expr)}"))
display(Math(rf"v = {sp.latex(v_expr)}"))
display(Math(rf"t = {sp.latex(t_expr)}"))
```

```python
# --- 1. Create Lambdified Functions from Cell 1 Expressions ---
calc_a = sp.lambdify((v, v0, x), a_expr)
calc_v = sp.lambdify((a, x, v0), v_expr)
calc_t = sp.lambdify((v0, a, x), t_expr)


# --- 2. Astropy Physical Inputs ---
v0_qty = 0 * (u.m / u.s)
v1_qty = 3.00 * (u.m / u.s)
x1_qty = 8.00 * u.m
x2_qty = 16.00 * u.m

# --- 4. Evaluate Lambdified Functions ---
a_qty = calc_a(v1_qty, v0_qty, x1_qty)
v2_qty = calc_v(a_qty, x2_qty, v0_qty)
t2_qty = calc_t(v0_qty, a_qty, x2_qty)


print(f"Calculated constant a = {a_qty:.2f}")
print(f"Part (a) When block is at {x2_qty} the speed v2 = {v2_qty:.2f}")
print(f"Part (b) When block is at {x2_qty} the time t2 = {t2_qty:.2f}")
```

#### 2.60 
••• An underground train starts from rest at a station and accelerates
at a rate of 1.60 m/s2 for 14.0 s. It runs at constant speed for 70.0
s and slows down at a rate of 3.50 m/s2 until it stops at the next station.
Find the total distance covered.

```python
# Define symbolic variables
x, x0, v, v0, t = sp.symbols("x, x0, v, v0, t", real=True)

# Kinematic equations
pos_eq = sp.Eq(x, x0 + v0 * t + sp.Rational(1, 2) * a * t**2)
vel_eq = sp.Eq(v, v0 * t + a * t)
vel_const_eq = sp.Eq(v, x / t)  # Constant velocity
vel_sq_eq = sp.Eq(v**2, v0**2 + 2 * a * x)  # velocity squared equation

# Position x
x_expr = sp.solve(pos_eq, x)[0]

# Velocity v
v_expr = sp.solve(vel_eq, v)[0]

# Position x (constant velocity)
x_const_expr = sp.solve(vel_const_eq, x)[0]

# Position x (from vel_sq_eq)
x_sq_expr = sp.solve(vel_sq_eq, x)[0]

# print results

print("First leg:")
display(Math(rf"{sp.latex(pos_eq)}"))


print("Second leg:")
display(Math(rf"{sp.latex(vel_eq)}"))
display(Math(rf"x = {sp.latex(x_const_expr)}"))


print("Tird leg:")
print("From the equation:")
display(Math(rf"{sp.latex(vel_sq_eq)}"))
print("Solve for x")
display(Math(rf"x = {sp.latex(x_sq_expr)}"))
```

```python
# --- 1. Create Lambdified Functions from Cell 1 Expressions ---
calc_x = sp.lambdify((a, t, v0, x0), x_expr)
calc_v = sp.lambdify((a, t, v0), v_expr)
calc_x_const = sp.lambdify((t, v), x_const_expr)
calc_x_sq_expr = sp.lambdify((v, v0, a), x_sq_expr)


# --- 2. Astropy Physical Inputs ---
v0_qty = 0 * (u.m / u.s)
x0_qty = 0 * u.m
a1_qty = 1.60 * (u.m / u.s**2)
t1_qty = 14 * u.s
t2_qty = 70.0 * u.s
a3_qty = -3.50 * (u.m / u.s**2)
v3_qty = 0 * (u.m / u.s)  # The train stops at next station


# --- 4. Evaluate Lambdified Functions ---
# --- First leg ---
x1_qty = calc_x(a1_qty, t1_qty, v0_qty, x0_qty)

# --- Second leg ---
v2_qty = calc_v(a1_qty.value, t1_qty.value, v0_qty.value) * (u.m / u.s)
x2_qty = calc_x_const(
    t2_qty,
    v2_qty,
)

# --- Tird leg --- (V=0, a3_qyt is negative )
x3_qty = calc_x_sq_expr(v3_qty, v2_qty, a3_qty)

print("=== Results ===")
print(f"In the first leg at time {t1_qty} the train has run x1 = {x1_qty:.2f}")
print(f"In the second leg the train runs at constant speed of v2 = {v2_qty:.2f}")
print(f"In the scond leg at time {t1_qty + t2_qty} the train has run x2 = {x2_qty:.2f}")
print(f"In the third leg the train has run x3 = {x3_qty:.2f}")
print(f"Total distance covered by the train x = {x1_qty + x2_qty + x3_qty:.2f}")
800701
```
